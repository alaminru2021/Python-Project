```python
# ============================================================
# SOCIOECONOMIC AND BEHAVIORAL DETERMINANTS
# OF ANXIETY AND DEPRESSION
# COMPLETE PYTHON ANALYSIS CODE
# ============================================================

# =========================
# 1. IMPORT LIBRARIES
# =========================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from scipy import stats

import statsmodels.api as sm
from statsmodels.formula.api import ols

from sklearn.linear_model import LinearRegression

# Optional reliability package
!pip install pingouin
from pingouin import cronbach_alpha
```

```python
# =========================
# 2. LOAD DATASET
# =========================

file_path = "Anxiety and Depression.xlsx"

df = pd.read_excel(file_path)

print(df.head())
```

```python
print(df.info())
```

```python
# =========================
# 3. CREATE COMPOSITE SCORES
# =========================

# Anxiety Score
gad_items = ['GAD1','GAD2','GAD3','GAD4','GAD5','GAD6','GAD7']

df['anxiety_score'] = df[gad_items].sum(axis=1)

# Depression Score
phq_items = ['PHQ1','PHQ2','PHQ3','PHQ4','PHQ5','PHQ6','PHQ7','PHQ8','PHQ9']

df['depression_score'] = df[phq_items].sum(axis=1)

print(df[['anxiety_score', 'depression_score']].head())
```

```python
# =========================
# 4. DESCRIPTIVE STATISTICS
# =========================

print("\nDescriptive Statistics")
print(df.describe())
```

```python
# =========================
# 5. FREQUENCY TABLES
# =========================

print("\nGender Frequency")
print(df['sex'].value_counts())

print("\nResidence Frequency")
print(df['residence'].value_counts())

print("\nPhysical Activity Frequency")
print(df['physical_activity'].value_counts())

print("\nAcademic Stress Frequency")
print(df['academic_stress'].value_counts())
```

```python
# =========================
# 6. RELIABILITY ANALYSIS
# =========================

# Anxiety Scale Reliability
alpha_gad = cronbach_alpha(df[gad_items])

print("\nCronbach Alpha - Anxiety Scale")
print(alpha_gad)

# Depression Scale Reliability
alpha_phq = cronbach_alpha(df[phq_items])

print("\nCronbach Alpha - Depression Scale")
print(alpha_phq)
```
