### Swapping

```python
# Wrong
a = 5
b = 10
a = b
b = a
print(a,b)  # a=10, b=10

#Correct
a = 5
b = 10
c = a
a = b
b = c
print(a,b)  # a=10, b=5

#Shortcut method
a = 5
b = 10
a, b = b, a
print(a,b)  # a=10, b=5
```

```python
# Temperature Conversion Menu

print("Temperature conversion menu:")
print("1. Celsius to Fahrenheit")
print("2. Fahrenheit to Celsius")
print("3. Celsius to Kelvin")
print("4. Kelvin to Celsius")
print("5. Fahrenheit to Kelvin")
print("6. Kelvin to Fahrenheit")

choice = int(input("Enter your choice: "))

if choice == 1:
    C = float(input("Enter the Celsius temperature: "))
    F = (C * 9 / 5) + 32
    print(f"The temperature in Fahrenheit is {F:.2f}")

elif choice == 2:
    F = float(input("Enter the Fahrenheit temperature: "))
    C = (F - 32) * 5 / 9
    print(f"The temperature in Celsius is {C:.2f}")

elif choice == 3:
    C = float(input("Enter the Celsius temperature: "))
    K = C + 273.15
    print(f"The temperature in Kelvin is {K:.2f}")

elif choice == 4:
    K = float(input("Enter the Kelvin temperature: "))
    C = K - 273.15
    print(f"The temperature in Celsius is {C:.2f}")

elif choice == 5:
    F = float(input("Enter the Fahrenheit temperature: "))
    K = (F - 32) * 5 / 9 + 273.15
    print(f"The temperature in Kelvin is {K:.2f}")

elif choice == 6:
    K = float(input("Enter the Kelvin temperature: "))
    F = (K - 273.15) * 9 / 5 + 32
    print(f"The temperature in Fahrenheit is {F:.2f}")

else:
    print("Not a correct option.")
```
